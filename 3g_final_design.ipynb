# Step 3, Part 1: final network design (v1.2)

The design on the Chattanooga-area grid: PQC on every link, QKD on 8 links chosen by the exact integer program with the balanced objective (secrecy behind QKD + tap detection on the busiest fiber, weight 0.25), at most 4 trusted nodes and chains of at most 3 hops, plus backup routes so no single fiber cut isolates a substation.

Writes `network_diagram.png` (used in WRITEUP.md) and `network_design/final_design.json`. Runs in under a minute.

```python
import importlib.util, subprocess, sys, json
for pkg in ("networkx", "scipy"):
    if importlib.util.find_spec(pkg) is None:
        %pip install -q {pkg}
from IPython.display import Image, display
r = subprocess.run([sys.executable, "network_design/final_design.py", "network_design/regions/chattanooga_r1.json",
                    "network_diagram.png", "network_design/final_design.json"], capture_output=True, text=True)
print(r.stdout or r.stderr)
display(Image("network_diagram.png", width=1000))
```

```python
d = json.load(open("network_design/final_design.json"))
print("QKD links (km):"); [print("  ", k, v) for k, v in d["qkd_link_km"].items()]
print("Trusted nodes:", d["trusted_nodes"])
print(f"Load behind QKD on every hop: {d['mw_behind_qkd']} of {d['total_mw']} MW")
b = d["backup"]; print(f"Single-cut weak spots: {b['bridges_before']} -> {b['bridges_after']}, worst load cut off {b['worst_cut_mw_before']} -> {b['worst_cut_mw_after']} MW, using {b['backup_links']} backup routes")
```
